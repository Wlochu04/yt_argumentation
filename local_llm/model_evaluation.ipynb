```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from local_lm import train_local_model

# Load
df = pd.read_excel('YT_comments.xlsx')
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)

# Train
model, vectorizer = train_local_model(train_df)

# Test
test_X = vectorizer.transform(test_df['comment'])
predictions = model.predict(test_X)

# Raport 
print(classification_report(test_df['category'], predictions))
```
